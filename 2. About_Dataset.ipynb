# **Dataset Information**

The dataset used in this project is downloaded from **Kaggle**. You can find the dataset [here](https://www.kaggle.com/datasets/abdallahwagih/telco-customer-churn?select=Telco_customer_churn.xlsx).

### **Dataset Overview**
This dataset contains detailed information about telecom customers in the United States, including their demographic details, subscription plans, and churn status. The data was collected to help identify the key factors contributing to customer churn, which occurs when customers cancel their services.

### **Key Attributes**
- **Customer Demographics**: Includes gender, city, state, zip code, and geographic coordinates (latitude and longitude).
- **Subscription Information**: This includes details about the customer's contract type (month-to-month, one-year, two-year), payment method (electronic check, mailed check, credit card, etc.), and monthly charges.
- **Churn Data**: Indicates whether the customer has churned (Yes/No), along with a score predicting their likelihood to churn, reasons for leaving, and the customer lifetime value (CLTV).
- **Churn Reasons**: Specific reasons for churn are included, such as competitors offering better deals, service dissatisfaction, or moving to a new location.

### **Key Columns**
- **CustomerID**: Unique identifier for each customer.
- **Gender**: Gender of the customer.
- **Contract**: The contract type (Month-to-month, One year, etc.).
- **Payment Method**: How the customer makes payments (e.g., Bank transfer, Credit card).
- **Monthly Charges**: The amount the customer is billed monthly.
- **Total Charges**: Cumulative amount billed to the customer.
- **Churn Label**: Indicates whether the customer has churned (Yes/No).
- **Churn Reason**: Reason why the customer churned.
  
### **Dataset Uses**
This dataset can be used for various machine learning tasks, including:
- **Customer Churn Prediction**: Identify patterns in customer behavior that signal their likelihood of churning.
- **Customer Segmentation**: Group customers based on their demographics, subscription plans, or payment behavior.
- **Churn Reason Analysis**: Understand the main reasons customers are leaving and build strategies to reduce churn.

### **Dataset Highlights**
- 33 features capturing various aspects of customer demographics, payment behavior, and churn data.
- Rich churn data with reasons for churn, which can help develop insights into customer retention strategies.
- Geospatial information, including customer locations (latitude, longitude), which can be useful for regional analysis of churn.
