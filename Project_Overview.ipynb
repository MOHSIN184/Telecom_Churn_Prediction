# **Customer Churn Prediction for a Telecom Company**

## **Project Overview**
This project focuses on predicting customer churn for a telecom company, using historical customer data. The goal is to identify customers who are likely to churn (cancel their service), so that the company can take proactive retention actions to reduce churn and improve customer retention.

### **Business Problem Statement**
The telecom company is experiencing customer loss (churn), which directly impacts revenue. The goal of this project is to identify customers who are likely to leave the service so that the company can take proactive retention actions, such as offering discounts, improving customer support, or better-targeted promotions.

### **Objective**
Build a machine learning model that predicts whether a customer will churn (Yes/No) based on customer demographics, service usage, and billing information.

### **Success Metrics**
- **Primary Metric**: **Recall for churned customers** (True Positive Rate)
  - **Reason**: Missing a churned customer (False Negative) costs more than contacting a loyal one.
- **Secondary Metrics**: 
  - **Precision** (to ensure that we don't contact customers who are unlikely to churn).
  - **F1-score** (a balance between precision and recall).


### **Assumptions**
1. **Historical Data Relevance**: Historical customer behavior patterns are indicative of future churn trends.
2. **Data Quality**: The dataset is accurate, complete, and free from significant biases or measurement errors.
3. **Behavioral Consistency**: Customers exhibiting similar demographic and behavioral characteristics will have comparable churn propensities.
4. **Feature Stability**: The relationships between features and churn outcomes remain relatively stable over time.
5. **Independent Observations**: Each customer record is independent and not influenced by other customers' decisions.

- Historical customer behavior reflects future churn patterns.
- The data is accurate, and there are no significant biases.
- Customers with similar behavior will churn similarly.

### **Business Impact**
By accurately identifying potential churners, the company can take proactive retention actions such as personalized offers, improved customer support, and targeted campaigns. This will lead to a reduced churn rate and increased customer lifetime value, directly contributing to higher revenue and customer loyalty.
